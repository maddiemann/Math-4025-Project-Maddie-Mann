### Notes about features 
- Aneuploidy score: measures the total number of chromosome arm gains or losses within a tumor. Ranges from 0-39. A higher score indicates more "chaos" in the genome and often more aggressive tumors. 
- Birth_from_Initial_Pathologic_Diagnosis_Date: Time interval from a person's date of birth to the date of initial pathologic diagnosis, represented as a calculated number of days.

New_Neoplasm_Event_Post_Initial_Therapy_Indicator: Time interval from a person's date of birth to the date of initial pathologic diagnosis, represented as a calculated number of days.

Neoplasm_Disease_Lymph_Node_Stage_American_Joint_Committee_on_Cancer_Code: The codes that represent the stage of cancer based on the nodes present (N stage) according to criteria based on multiple editions of the AJCC's Cancer Staging Manual.

Buffa_Hypoxia_Score: a molecular tool used to quantify the level of oxygen deprivation (hypoxia) within tumors. Measures the activity of genes regulated by low oxygen, high scores are associated with poorer prognosis 

Ragnum_Hypoxia_Score: Measures oxygenation levels particularly in prostate cancer.

Disease_Free_(Months): Amount of time in remission. Likely will be predictor variable
